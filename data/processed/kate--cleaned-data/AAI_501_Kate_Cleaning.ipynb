```python
import pandas as pd

# Load the original dataset
df = pd.read_csv("aisecwatch-dataset-2026-10-10.csv")

# Check dataset dimensions
print("Dataset shape:", df.shape)

# Display the first five rows
display(df.head())
```

```python
# Check missing values in each column
missing_values = df.isna().sum()

# Sort columns by the number of missing values
missing_values = missing_values.sort_values(ascending=False)

display(missing_values)
```

```python
# Check severity distribution
print(df["severity"].value_counts())

# Check the sources of severity labels
print(df["severity_source"].value_counts())
```

```python
# Step 1: Keep only CVSS-based severity labels
# with the four target classes

valid_classes = ["low", "medium", "high", "critical"]

df_clean = df[
    (df["severity_source"] == "cvss") &
    (df["severity"].isin(valid_classes))
].copy()

# Check the result
print("Original dataset:", df.shape)
print("Filtered dataset:", df_clean.shape)

print("\nSeverity distribution:")
print(df_clean["severity"].value_counts())
```

```python
# Check missing values in the filtered dataset
missing_after = df_clean.isna().sum()

missing_after = missing_after[
    missing_after > 0
].sort_values(ascending=False)

display(missing_after.to_frame("Missing Values"))
```

```python
# Step 3: Check for duplicates

# Exact duplicate rows
print("Exact duplicate rows:", df_clean.duplicated().sum())

# Duplicate record IDs
print("Duplicate IDs:", df_clean["id"].duplicated().sum())

# Duplicate titles
print("Duplicate titles:", df_clean["title"].duplicated().sum())

# Duplicate title and summary combinations
print(
    "Duplicate title + summary:",
    df_clean.duplicated(subset=["title", "summary"]).sum()
)

# Duplicate non-missing CVE IDs
cve_ids = df_clean["cve_id"].dropna()

print("Duplicate CVE IDs:", cve_ids.duplicated().sum())
```

**Duplicate Detection:** The dataset was checked for exact duplicate rows, duplicate record IDs, duplicate titles, duplicate title-summary combinations, and duplicate non-null CVE IDs. No duplicates were found using these criteria, so no records were removed during this step.

```python
# Step 4: Inspect records with missing summaries

missing_summary = df_clean[df_clean["summary"].isna()]

display(
    missing_summary[
        ["id", "title", "summary", "severity"]
    ]
)

# Check whether any records have both text fields missing
missing_both = (
    df_clean["title"].isna() &
    df_clean["summary"].isna()
).sum()

print("Missing both title and summary:", missing_both)
```

```python
# Step 5: Handle missing text values

# Replace missing text values with empty strings
df_clean["title_clean"] = df_clean["title"].fillna("").str.strip()
df_clean["summary_clean"] = df_clean["summary"].fillna("").str.strip()

# Combine title and summary into one text column
df_clean["model_text"] = (
    df_clean["title_clean"] + " " + df_clean["summary_clean"]
).str.strip()

# Check the result
print("Dataset shape:", df_clean.shape)
print("Missing model_text:", df_clean["model_text"].isna().sum())
print("Empty model_text:", (df_clean["model_text"] == "").sum())

# Display the two records that originally had missing summaries
display(
    df_clean.loc[
        missing_summary.index,
        ["title_clean", "summary_clean", "model_text", "severity"]
    ]
)
```

**Missing Text Values:** Two records had missing summary values, but both contained valid titles. Instead of dropping these records, missing summaries were replaced with empty strings in a separate cleaned column. The title and summary fields were combined into a new "model_text" feature. This preserved all 2,444 records for further preprocessing.

```python
# Step 6: Identify potential data leakage

# Compare the target with CVSS severity
matching_labels = (
    df_clean["severity"] == df_clean["cvss_severity"]
).sum()

print("Total records:", len(df_clean))
print("Matching severity labels:", matching_labels)

# Check CVSS score availability
print(
    "Records with CVSS score:",
    df_clean["cvss_score"].notna().sum()
)

# Check whether priority provides useful variation
print(
    "Unique priority values:",
    df_clean["priority"].unique()
)

# Check whether source_category varies
print(
    "Unique source categories:",
    df_clean["source_category"].unique()
)
```

```python
# Step 7: Detect potential data leakage in text

# Search for CVSS references and explicit severity labels
leakage_pattern = (
    r"\bcvss\b|"
    r"\bcvssv?\d+(?:\.\d+)?\b|"
    r"\b(?:critical|high|medium|low)\s+severity\b|"
    r"\bseverity\s*[:=-]\s*(?:critical|high|medium|low)\b"
)

# Identify potentially problematic records
potential_leakage = df_clean[
    df_clean["model_text"].str.contains(
        leakage_pattern,
        case=False,
        regex=True,
        na=False
    )
].copy()

print("Total records:", len(df_clean))
print("Potential leakage records:", len(potential_leakage))

print(
    "Percentage:",
    round(len(potential_leakage) / len(df_clean) * 100, 2),
    "%"
)

# Display examples for manual inspection
potential_leakage["text_preview"] = (
    potential_leakage["model_text"].str.slice(0, 350)
)

display(
    potential_leakage[
        ["id", "severity", "text_preview"]
    ].head(10)
)
```

**Data Leakage Assessment**

The filtered dataset contains 2,444 records. The "cvss_severity" field matches the target variable ("severity") in all 2,444 records, confirming direct target leakage. Additionally, "cvss_score" is available for 1,869 records and is directly related to severity classification. These fields will be excluded from model input features.

The "priority" and "source_category" fields contain only one unique value each in the filtered dataset and therefore provide no discriminatory information.

Text descriptions will also be examined for explicit CVSS scores and severity ratings before model training.

```python
# Step 8: Inspect potential leakage in context

import re

# Extract the surrounding text for each regex match
def get_leakage_context(text):
    snippets = []

    for match in re.finditer(
        leakage_pattern, text, flags=re.IGNORECASE
    ):
        start = max(0, match.start() - 80)
        end = min(len(text), match.end() + 120)

        snippets.append(
            text[start:end].replace("\n", " ")
        )

    return " | ".join(snippets[:2])


# Apply the function to flagged records
potential_leakage["leakage_context"] = (
    potential_leakage["model_text"].apply(get_leakage_context)
)

# Show longer text in the output
pd.set_option("display.max_colwidth", 300)

display(
    potential_leakage[
        ["severity", "leakage_context"]
    ].head(15)
)
```

```python
# Step 9: Remove potentially leaking sentences

def remove_flagged_sentences(text):
    # Split summary into sentences
    sentences = re.split(r'(?<=[.!?])\s+', text)

    # Keep sentences without the leakage pattern
    clean_sentences = [
        sentence for sentence in sentences
        if not re.search(
            leakage_pattern,
            sentence,
            flags=re.IGNORECASE
        )
    ]

    return " ".join(clean_sentences).strip()


# Create a new cleaned summary
df_clean["summary_no_ratings"] = (
    df_clean["summary_clean"].apply(remove_flagged_sentences)
)

# Combine the title with the cleaned summary
df_clean["model_text_candidate"] = (
    df_clean["title_clean"] + " " +
    df_clean["summary_no_ratings"]
).str.strip()

# Check the cleaning results
changed = (
    df_clean["summary_clean"] !=
    df_clean["summary_no_ratings"]
)

print("Modified summaries:", changed.sum())

print(
    "Empty model texts:",
    (df_clean["model_text_candidate"] == "").sum()
)

print(
    "Summaries now empty:",
    (df_clean["summary_no_ratings"] == "").sum()
)

# Compare original and cleaned descriptions
for idx in df_clean.index[changed][:3]:
    print("\n" + "=" * 60)
    print("Record:", idx)
    print("BEFORE:")
    print(df_clean.loc[idx, "summary_clean"])
    print("\nAFTER:")
    print(df_clean.loc[idx, "summary_no_ratings"])
```

```python
# Step 10: More precise removal of severity ratings

def redact_ratings(text):
    patterns = [
        # CVSS vectors
        r"\bCVSS:\s*[234](?:\.\d+)?(?:/[A-Za-z0-9:]+)+",

        # CVSS scores and optional severity labels
        r"\bCVSS(?:[- ]?B)?\s*(?:v?[234]\.\d+\s+)?"
        r"(?:base\s+)?(?:score\s*(?:is|of|around)?\s*)?"
        r"~?\s*\d{1,2}(?:\.\d+)?"
        r"(?:\s*\(\s*(?:critical|high|medium|low)\s*\)"
        r"|\s+(?:critical|high|medium|low)\b)?",

        # Explicit severity-rating phrases
        r"\b(?:rated|rates?\s+it|rates?\s+the\s+issue|"
        r"classified\s+as|severity\s*(?:of|[:=-]))"
        r"\s*(?:as\s+)?(?:critical|high|medium|low)\b",

        # Phrases such as "high severity"
        r"\b(?:critical|high|medium|low)[- ]severity\b"
    ]

    for pattern in patterns:
        text = re.sub(pattern, "", text, flags=re.IGNORECASE)

    return re.sub(r"\s+", " ", text).strip()


# Start again from the original cleaned summary
df_clean["summary_redacted_v2"] = (
    df_clean["summary_clean"].apply(redact_ratings)
)

# Compare text loss
for idx in [351, 353, 1734]:
    original = df_clean.loc[idx, "summary_clean"]
    revised = df_clean.loc[idx, "summary_redacted_v2"]

    print(f"\nRecord {idx}")
    print("Characters before:", len(original))
    print("Characters after:", len(revised))
    print("\nREVISED TEXT:\n", revised)
    print("-" * 60)

print(
    "\nEmpty summaries:",
    (df_clean["summary_redacted_v2"] == "").sum()
)
```

```python
# Step 11: Fix punctuation and wording artifacts

def polish_text(text):
    # Remove empty parentheses
    text = re.sub(r"\(\s*\)", "", text)

    # Fix "rated and mapped"
    text = re.sub(
        r"\brated\s+and\b",
        "and",
        text,
        flags=re.IGNORECASE
    )

    # Fix "The flaw is , , and allows..."
    text = re.sub(
        r"\bis\s*,\s*,\s*and\b",
        "",
        text,
        flags=re.IGNORECASE
    )

    # Remove repeated commas and extra spaces
    text = re.sub(r"\s*,\s*,", ",", text)
    text = re.sub(r"\s+([,.:;])", r"\1", text)
    text = re.sub(r"\s+", " ", text)

    return text.strip()


# Create version 3
df_clean["summary_redacted_v3"] = (
    df_clean["summary_redacted_v2"].apply(polish_text)
)

df_clean["model_text_clean"] = (
    df_clean["title_clean"] + " " +
    df_clean["summary_redacted_v3"]
).str.strip()

# Display corrected examples
for idx in [351, 353, 1734]:
    print(f"\nRecord {idx}:")
    print(df_clean.loc[idx, "summary_redacted_v3"])
    print("-" * 60)

# Check for remaining potential leakage
remaining_leakage = df_clean["model_text_clean"].str.contains(
    leakage_pattern,
    case=False,
    regex=True,
    na=False
)

print("\nRemaining flagged records:", remaining_leakage.sum())
print("Empty model texts:", (df_clean["model_text_clean"] == "").sum())
```

```python
# Step 12: Final punctuation corrections
# Preserve technical expressions such as eval()

def polish_text_safe(text):

    # Fix empty parentheses in the specific rating phrase
    text = re.sub(
        r"\bthe lead one\s*\(\s*\)\s*:",
        "the lead one:",
        text,
        flags=re.IGNORECASE
    )

    # Fix "rated and mapped to CWE"
    text = re.sub(
        r",\s*rated\s+and\s+mapped\s+to\b",
        " and is mapped to",
        text,
        flags=re.IGNORECASE
    )

    # Fix "The flaw is , , and allows"
    text = re.sub(
        r"\bThe flaw is\s*,\s*,\s*and\s+allows\b",
        "The flaw allows",
        text,
        flags=re.IGNORECASE
    )

    # Remove unnecessary spaces
    text = re.sub(r"\s+([,.:;])", r"\1", text)
    text = re.sub(r"\s{2,}", " ", text)

    return text.strip()


# Rebuild from version 2, not version 3
df_clean["summary_redacted_v3"] = (
    df_clean["summary_redacted_v2"].apply(polish_text_safe)
)

df_clean["model_text_clean"] = (
    df_clean["title_clean"] + " " +
    df_clean["summary_redacted_v3"]
).str.strip()

# Verify important technical expressions
for idx in [351, 353, 1734]:
    print(f"\nRecord {idx}:")
    print(df_clean.loc[idx, "summary_redacted_v3"])
    print("-" * 60)

print(
    "\nEmpty model texts:",
    (df_clean["model_text_clean"] == "").sum()
)
```

```python
# Step 13: Validate severity labels against CVSS scores

# Convert CVSS scores to numeric values
df_clean["cvss_score_numeric"] = pd.to_numeric(
    df_clean["cvss_score"],
    errors="coerce"
)

# Define the expected severity based on CVSS score
def get_cvss_severity(score):
    if pd.isna(score):
        return None
    elif score == 0:
        return "none"
    elif score < 4.0:
        return "low"
    elif score < 7.0:
        return "medium"
    elif score < 9.0:
        return "high"
    else:
        return "critical"


# Calculate severity from available CVSS scores
df_clean["expected_severity"] = (
    df_clean["cvss_score_numeric"].apply(get_cvss_severity)
)

# Find records where the labels disagree
mismatches = df_clean[
    df_clean["expected_severity"].notna() &
    (df_clean["severity"] != df_clean["expected_severity"])
].copy()

print("Total records:", len(df_clean))
print("Records with CVSS scores:",
      df_clean["cvss_score_numeric"].notna().sum())
print("Severity mismatches:", len(mismatches))

display(
    mismatches[
        ["id", "cvss_score", "cvss_vector",
         "severity", "expected_severity"]
    ]
)
```

```python
# Step 14: Correct severity labels (reproducible version)

# Verify that original record IDs still match
assert (
    df.loc[df_clean.index, "id"] == df_clean["id"]
).all()

# Always retrieve original labels from raw data
df_clean["severity_original"] = (
    df.loc[df_clean.index, "severity"].copy()
)

# Reset target to original values before corrections
df_clean["severity"] = df_clean["severity_original"].copy()

# Identify confirmed inconsistencies
correction_mask = (
    df_clean["expected_severity"].notna() &
    df_clean["severity"].ne(df_clean["expected_severity"])
)

# Document corrections
df_clean["cleaning_note"] = ""

df_clean.loc[correction_mask, "cleaning_note"] = (
    "Severity corrected from "
    + df_clean.loc[correction_mask, "severity"]
    + " to "
    + df_clean.loc[correction_mask, "expected_severity"]
    + " based on validated CVSS score"
)

# Apply corrections
df_clean.loc[correction_mask, "severity"] = (
    df_clean.loc[correction_mask, "expected_severity"]
)

print("Corrected records:", correction_mask.sum())
print(
    "Original labels preserved:",
    df_clean["severity_original"].notna().sum()
)
```

### Severity Label Validation and Correction

Severity labels were validated against available numerical CVSS scores using the standard severity thresholds. Ten inconsistencies were identified among the 1,869 records containing numerical CVSS scores.

The corresponding CVSS 3.1 vectors were independently checked, confirming that the numerical scores were consistent with the vector calculations. Eight records were corrected from Low to Medium, and two records were corrected from Medium to High.

Original labels were preserved in the "severity_original" column, and each correction was documented in "cleaning_note" to ensure traceability. All 2,444 records were retained.

The correction cell is idempotent: it resets labels from the source data before reapplying the ten validated fixes.\n

```python
# Step 15: Restore meaningful technical information

# Start from our latest text-cleaning version
df_clean["summary_v4"] = df_clean["summary_redacted_v3"].copy()

# Find descriptions mentioning high attack complexity
complexity_mask = df_clean["summary_clean"].str.contains(
    "rated high in complexity",
    case=False,
    regex=False,
    na=False
)

print("Records to inspect:", complexity_mask.sum())

# Restore the technical meaning
for idx in df_clean.index[complexity_mask]:
    df_clean.at[idx, "summary_v4"] = re.sub(
        r"\bis\s+in\s+complexity\b",
        "is rated high in complexity",
        df_clean.at[idx, "summary_v4"],
        flags=re.IGNORECASE
    )

# Rebuild the updated model text
df_clean["model_text_v4"] = (
    df_clean["title_clean"] + " " +
    df_clean["summary_v4"]
).str.strip()

# Compare original and updated descriptions
for idx in df_clean.index[complexity_mask]:
    print("\nRecord:", idx)
    print("\nORIGINAL:")
    print(df_clean.at[idx, "summary_clean"])

    print("\nUPDATED:")
    print(df_clean.at[idx, "summary_v4"])

print("\nTotal records:", len(df_clean))
print(
    "Empty model texts:",
    (df_clean["model_text_v4"] == "").sum()
)
```

```python
# Step 16: Final text leakage audit

# Patterns that may indicate explicit severity ratings
audit_pattern = re.compile(
    r"\bcvss\b|"
    r"\b(?:critical|high|medium|low)[- ]severity\b|"
    r"\b(?:severity|rating)\s*(?:is|was|of|[:=])\s*"
    r"(?:critical|high|medium|low)\b|"
    r"\b(?:rated|classified as|assigned)\s+"
    r"(?:as\s+)?(?:critical|high|medium|low)\b",
    flags=re.IGNORECASE
)

# Find records requiring manual inspection
audit_mask = df_clean["model_text_v4"].str.contains(
    audit_pattern,
    na=False
)

print("Total records:", len(df_clean))
print("Records flagged for review:", audit_mask.sum())

# Extract the context around matches
def extract_audit_context(text):
    matches = list(audit_pattern.finditer(text))

    contexts = []
    for match in matches[:2]:
        start = max(0, match.start() - 70)
        end = min(len(text), match.end() + 100)
        contexts.append(text[start:end])

    return " | ".join(contexts)

audit_results = df_clean.loc[
    audit_mask, ["severity", "model_text_v4"]
].copy()

audit_results["context"] = (
    audit_results["model_text_v4"]
    .apply(extract_audit_context)
)

pd.set_option("display.max_colwidth", 250)

display(
    audit_results[
        ["severity", "context"]
    ].head(20)
)
```

```python
# Step 17: Remove remaining explicit severity labels

# CVE records with explicit severity wording
problem_cves = [
    "CVE-2025-6855",
    "CVE-2025-6853",
    "CVE-2025-3677",
    "CVE-2025-3001",
    "CVE-2025-3000",
    "CVE-2025-2999",
    "CVE-2025-2733",
    "CVE-2024-3078",
    "CVE-2024-2057"
]

def remove_explicit_rating(text):
    labels = r"(?:critical|high|medium|low)"

    # Remove standalone rating statements
    text = re.sub(
        rf"\bIt\s+(?:has\s+been|was|is)\s+"
        rf"(?:rated|classified)\s+as\s+{labels}\s*\.\s*",
        "",
        text,
        flags=re.IGNORECASE
    )

    # Remove "which was classified as critical"
    text = re.sub(
        rf",?\s+which\s+(?:was|is|has\s+been)\s+"
        rf"(?:classified|rated)\s+as\s+{labels}\s*,?",
        " ",
        text,
        flags=re.IGNORECASE
    )

    # Remove "classified as critical"
    text = re.sub(
        rf"\b(?:and\s+)?(?:was\s+|is\s+|has\s+been\s+)?"
        rf"classified\s+as\s+{labels}\b",
        "",
        text,
        flags=re.IGNORECASE
    )

    # Remove "rated critical"
    text = re.sub(
        rf"\brated\s+(?:as\s+)?{labels}\b",
        "",
        text,
        flags=re.IGNORECASE
    )

    # Remove severity adjectives before vulnerability/flaw
    text = re.sub(
        rf"\b{labels}\s+"
        rf"(?=(?:[\w-]+\s+){{0,4}}(?:vulnerability|flaw)\b)",
        "",
        text,
        flags=re.IGNORECASE
    )

    # Fix spacing and punctuation
    text = re.sub(r"\s+([.,;:])", r"\1", text)
    text = re.sub(r",\s*,", ",", text)
    text = re.sub(r"\s{2,}", " ", text)

    return text.strip()


# Preserve the previous text version
df_clean["title_v5"] = df_clean["title_clean"].copy()
df_clean["summary_v5"] = df_clean["summary_v4"].copy()

# Select only the verified records
problem_mask = df_clean["cve_id"].isin(problem_cves)

print("Records selected:", problem_mask.sum())

assert problem_mask.sum() == 9

# Apply targeted corrections
for col in ["title_v5", "summary_v5"]:
    df_clean.loc[problem_mask, col] = (
        df_clean.loc[problem_mask, col]
        .apply(remove_explicit_rating)
    )

# Rebuild model text
df_clean["model_text_v5"] = (
    df_clean["title_v5"] + " " +
    df_clean["summary_v5"]
).str.strip()

# Verify changes
print(
    "Modified model texts:",
    (df_clean["model_text_v5"] !=
     df_clean["model_text_v4"]).sum()
)

print(
    "Empty model texts:",
    (df_clean["model_text_v5"] == "").sum()
)

print(
    "Remaining audit flags:",
    df_clean["model_text_v5"].str.contains(
        audit_pattern, na=False
    ).sum()
)

# Display two corrected examples
for cve in ["CVE-2025-6855", "CVE-2025-6853"]:
    row = df_clean[df_clean["cve_id"] == cve].iloc[0]
    print("\n", cve)
    print("BEFORE:", row["model_text_v4"])
    print("AFTER:", row["model_text_v5"])
```

### Step 24: Residual Leakage Audit and Targeted Text Repair

An expanded audit found additional source-specific severity statements and CVSS score fragments that the initial regular expressions missed. A conservative follow-up removed explicit overall-severity ratings and repaired malformed sentences while preserving exploit mechanisms, software functions, CWE identifiers, and attack-complexity descriptions. In particular, words such as **HIGH** and **CRITICAL** used for *prompt-injection rule thresholds* are retained because they describe application behavior rather than the target vulnerability severity. Source metadata and original labels remain in the export for audit, but must not be used as model features.

This regex- and sample-based audit reduces known leakage; it cannot prove that all implicit rating information or train/test similarity is absent.

```python
# Step 24: Residual severity-leakage cleanup and text repair
# The previous audit only looked for a limited set of rating expressions.
# This pass targets leftover explicit labels/scores and repairs verified
# artifacts; it never changes target labels, IDs, chronological split, or rows.

import re

SEVERITY_WORDS = r"(?:critical|high|medium|low)"

def clean_residual_ratings(text):
    # Remove explicit overall severity classifications (not technical CWE names,
    # low-privilege attackers, high attack complexity, or rule thresholds).
    text = re.sub(
        r"\bIt has been declared as (?:critical|high|medium|low)\.\s*",
        "", text, flags=re.IGNORECASE
    )
    text = re.sub(
        r"\bThe source rates the severity (?:critical|high|medium|low)\.\s*",
        "", text, flags=re.IGNORECASE
    )
    text = re.sub(
        r"\bcritical\s+(?=(?:remote\s+)?(?:code execution|OS command injection|SQL injection|authentication bypass|vulnerability|flaw|RCE|API endpoints)\b)",
        "", text, flags=re.IGNORECASE
    )
    # Phrases indicating source-specific *overall* ratings, not technical detail.
    text = re.sub(
        r"\bThe advisory rates this (?:critical|high|medium|low) and notes that\b",
        "The advisory notes that", text, flags=re.IGNORECASE
    )
    text = re.sub(
        r"\bMaintainers rate the severity as (?:critical|high|medium|low), noting\b",
        "Maintainers note", text, flags=re.IGNORECASE
    )
    text = re.sub(
        r"\bThe source rates the risk as (?:critical|high|medium|low) for\b",
        "Affected deployment contexts include", text, flags=re.IGNORECASE
    )
    text = re.sub(
        r"\bThe source rates the confidentiality impact as (?:critical|high|medium|low), since\b",
        "The confidentiality risk arises because", text, flags=re.IGNORECASE
    )
    # Remove a score that survived the original CVSS regex.
    text = re.sub(
        r"\bThe estimated score is around\s*\d{1,2}(?:\.\d+)?\s*"
        r"\(\s*(?:critical|high|medium|low)\s*\)\s*for the local, agent-mediated case\.",
        "The flaw can be exploited by an agent operating locally.", text,
        flags=re.IGNORECASE
    )
    text = re.sub(
        r"\bThe base score from CNA VulDB is\s*\d{1,2}(?:\.\d+)?\s*"
        r"\(\s*(?:critical|high|medium|low)\s*\),\s*and\s*",
        "", text, flags=re.IGNORECASE
    )
    text = re.sub(
        r"\bThe, and the v4\.0 score is\s*\d+(?:\.\d+)?\s*"
        r"\(\s*(?:critical|high|medium|low)\s*\)\s*per the first\.org calculator,\s*"
        r"while GitHub's calculator returns\s*\d+(?:\.\d+)?\s*"
        r"\(\s*(?:critical|high|medium|low)\s*\)\.",
        "", text, flags=re.IGNORECASE
    )
    text = re.sub(
        r"\bThe CNA, GitHub, Inc\., rates it\s*\d+(?:\.\d+)?\s*"
        r"(?:critical|high|medium|low) under with a network attack vector, "
        r"no privileges required, and a low availability impact\.",
        "The attack is network-based and requires no privileges.",
        text, flags=re.IGNORECASE
    )
    # Residual descriptions of numeric ratings that are no longer complete.
    text = re.sub(r"\bThe source under \(GitHub, Inc\., base score 6\.9\)\.", "", text, flags=re.IGNORECASE)
    text = re.sub(r"\bGitHub, Inc\. assigned a, while NIST has not yet provided an assessment\.", "", text, flags=re.IGNORECASE)
    text = re.sub(r"\bThe base score from VulDB is, and NIST has not yet provided an assessment\.", "", text, flags=re.IGNORECASE)
    text = re.sub(r"\bVulnCheck rates (?:it|the issue)\s*(?:\(\)|, with a vector of [A-Z0-9:/]+)?"
                  r"\s*[,;]\s*(?:while|and) N(?:IST|VD) has not yet provided an (?:NVD )?assessment\.",
                  "", text, flags=re.IGNORECASE)
    text = re.sub(r"\bVulnCheck rates (?:it|the issue)\s*(?:\(\)|, with a vector of [A-Z0-9:/]+)?"
                  r"\s*[,;]\s*(?:with )?NVD assessment not yet provided\.",
                  "", text, flags=re.IGNORECASE)
    text = re.sub(r"\bVulDB rates it\s*(?:\([A-Z0-9:/]+\))?\s*"
                  r"[,;]\s*(?:while|with) N(?:IST|VD) (?:assessment not yet provided|has not yet provided an assessment)\.",
                  "", text, flags=re.IGNORECASE)
    text = re.sub(r"\bVulDB rates it\s*(?:\([A-Z0-9:/]+\))?\s*"
                  r"[,;]\s*N(?:IST|VD) (?:assessment not yet provided|has not yet provided an assessment)\.",
                  "", text, flags=re.IGNORECASE)
    text = re.sub(r"\bVulDB rated it\s*\([A-Z0-9:/]+\)", "", text, flags=re.IGNORECASE)
    text = re.sub(r"\bVulDB scores it, while NIST has not yet provided an assessment\.", "", text, flags=re.IGNORECASE)
    text = re.sub(r"\bVulnCheck rates it\s*[,;]\s*while NVD has not yet provided an assessment\.", "", text, flags=re.IGNORECASE)
    # Standalone fragments left when the CVSS score was stripped earlier.
    text = re.sub(r"\b(?:VulnCheck|VulDB|JFrog|GitHub(?:, Inc\.)?) rates (?:it|the issue)(?: \(\))?\.", "", text, flags=re.IGNORECASE)
    text = re.sub(r"\bThe source rates it\.", "", text, flags=re.IGNORECASE)
    text = re.sub(r"\bThe advisory rates (?:it|this)\.", "", text, flags=re.IGNORECASE)
    text = re.sub(r"\bThe source rates it and says\b", "The source says", text, flags=re.IGNORECASE)
    text = re.sub(r"\bThe advisory rates it and affects\b", "The advisory affects", text, flags=re.IGNORECASE)
    text = re.sub(r"\bThe CNA, (?:AMZN|GitHub, Inc\.), rates it, and N(?:IST|VD) has not yet provided an assessment\.", "", text, flags=re.IGNORECASE)
    text = re.sub(r"\bThe source rates it from VulnCheck, with NIST's assessment not yet provided\.", "", text, flags=re.IGNORECASE)
    text = re.sub(r"\bThe advisory rates it and notes that\b", "The advisory notes that", text, flags=re.IGNORECASE)
    text = re.sub(r"\bThe source rates the issue under CWE-200\b", "The issue is classified under CWE-200", text, flags=re.IGNORECASE)
    # Remove rating-only clauses attached to a substantive first clause.
    text = re.sub(
        r",\s+and (?:VulnCheck|VulDB) rates (?:it|the issue),\s*"
        r"(?:while|and) N(?:IST|VD) has not yet provided an assessment",
        "", text, flags=re.IGNORECASE
    )
    text = re.sub(r"\s+([,.;:])", r"\1", text)
    text = re.sub(r"\s{2,}", " ", text)
    return text.strip()

# Specific repairs anchored to verified source descriptions. These preserve
# technical distinctions that generic severity regexes cannot safely infer.
verified_replacements = {
    "CVE-2026-83071": [
        ("The, with high confidentiality, integrity and availability impacts.",
         "The compromise affects confidentiality, integrity, and availability."),
    ],
    "CVE-2026-61439": [
        ("block threshold to, so HIGH-level threats", "block threshold to CRITICAL, so HIGH-level threats"),
        ("detected at and logged", "detected at HIGH severity and logged"),
        ("VulnCheck rates the issue, and NVD has not yet provided an assessment.", ""),
    ],
    "CVE-2026-69255": [("The report also lists credential exposure via FLOWISE_PASSWORD and an arbitrary file read, with a.",
                          "The report also lists credential exposure via FLOWISE_PASSWORD and arbitrary file read.")],
    "CVE-2026-47411": [("an attacker could alter configuration such as provider URLs or invite behaviour, with a.",
                          "an attacker could alter configuration such as provider URLs or invite behaviour.")],
    "CVE-2025-62609": [
        ("The CNA, GitHub, Inc., rates it 5.5 MEDIUM under with a network attack vector, no privileges required, and a low availability impact.",
         "The attack is network-based and requires no privileges."),
    ],
    "CVE-2025-0649": [("Google Inc. rated it, with network attack vector, no privileges or user interaction required, and high availability impact.",
                         "The attack is network-based and requires no privileges or user interaction."),
    ],
    "CVE-2026-62186": [("VulnCheck rates it, with a vector of AV:N/AC:L/AT:N/PR:L/UI:N, and NIST has not yet provided an assessment.", "")],
    "CVE-2026-73614": [("VulnCheck rates it (); NVD has not yet provided an assessment.", "")],
    "CVE-2025-4701": [("VulDB rates it (AV:L/AC:L/AT:N/PR:L/UI:N/VC:L/VI:L/VA:L/SC:N/SI:N/SA:N), with NIST assessment not yet provided.", "")],
    "CVE-2026-14647": [("NVD had not yet provided an assessment, and VulDB rated it (AV:N/AC:L/AT:N/PR:L/UI:N/VC:L/VI:N/VA:N/SC:N/SI:N/SA:N/E:P).",
                          "NVD had not yet provided an assessment.")],
    "CVE-2026-33057": [("The advisory rates the impact as trivial severity for publicly exposed or unfirewalled deployments.",
                          "The impact is most relevant to publicly exposed or unfirewalled deployments.")],
    "CVE-2026-50027": [("The source rates it.", "")],
    "CVE-2026-47392": [("and the advisory rates it.", "" )],
    "CVE-2026-75104": [("The source rates it from VulnCheck, with NIST's assessment not yet provided.", "")],
    "CVE-2025-62372": [("and GitHub rates it, with availability impact only.",
                          "and the impact is limited to availability.")],
    "CVE-2026-35651": [("and VulnCheck rates it, with NVD assessment not yet provided.", "")],
    "CVE-2026-7061": [("VulDB rates it; NIST has not yet provided an assessment.", "")],
    "CVE-2026-6608": [("VulDB rates it, while NVD has not yet provided an assessment.", "")],
    "CVE-2026-4538": [("while VulDB rates it.", "")],
    "CVE-2026-1953": [],
    "CVE-2025-2148": [("VulDB scores it, while NIST has not yet provided an assessment.", "")],
    "CVE-2026-44220": [("The, and the v4.0 score is 5.7 (Medium) per the first.org calculator, while GitHub's calculator returns 2.4 (Low).", "")],
}

def apply_verified_repairs(text, cve):
    for before, after in verified_replacements.get(cve, []):
        if before in text:
            text = text.replace(before, after)
    return text

# Restore source-specific *prompt-injection filter configuration values* before
# rating cleanup; 'CRITICAL'/'HIGH' describe rule thresholds, NOT CVE severity.
df_clean["model_text_final"] = [
    clean_residual_ratings(apply_verified_repairs(t, str(cve)))
    for t, cve in zip(df_clean["model_text_v7"], df_clean["cve_id"].fillna(""))
]

# Remove orphaned boilerplate and smooth punctuation without altering syntax
# in paths, filenames, Python functions, and CVSS-independent exploit details.
df_clean["model_text_final"] = (
    df_clean["model_text_final"]
    .str.replace(r"\s+([,.;:])", r"\1", regex=True)
    .str.replace(r"\s{2,}", " ", regex=True)
    .str.strip()
)

# Final source-verified repairs for residual score vectors, rating-only
# sentences, and a few broken endings missed by initial patterns.
last_repairs = {
    "CVE-2025-43850": [("The vector from GitHub, Inc. is AV:N/AC:L/AT:N/PR:N/UI:N/VC:H/VI:H/VA:H, and NIST has not yet provided an assessment.", "")],
    "CVE-2026-10814": [("its complexity is, and exploitability", "its complexity is high, and exploitability")],
    "CVE-2026-45397": [("The advisory, with a.", "")],
    "CVE-2026-6662": [("VulDB scores it, with weaknesses CWE-346 and CWE-942.", "The weaknesses are CWE-346 and CWE-942.")],
    "CVE-2026-70473": [("The advisory rates this information disclosure.", "The advisory describes information disclosure.")],
    "CVE-2025-49841": [("GitHub, Inc. rated it, with network attack vector and no privileges or user interaction required.",
                          "The attack is network-based and requires no privileges or user interaction.")],
    "CVE-2025-32434": [("GitHub, Inc. scored it, with no NVD assessment yet.", "")],
    "CVE-2025-10155": [("JFrog rates it, and NVD has not yet provided an assessment.", "")],
    "CVE-2026-63766": [("VulnCheck scores it, while NVD has not yet provided an assessment.", "")],
    "CVE-2026-78684": [("VulnCheck rates it; NVD has not yet provided an assessment.", "")],
    "CVE-2026-59806": [("VulnCheck rates it; NVD has not yet provided an assessment.", "")],
    "CVE-2026-48545": [("VulnCheck rates it; NVD has not yet provided an assessment.", "")],
    "CVE-2024-14021": [("VulnCheck rates it; NVD has not yet provided an assessment.", "")],
}

for idx, row in df_clean.iterrows():
    text = df_clean.at[idx, "model_text_final"]
    for before, after in last_repairs.get(str(row["cve_id"]), []):
        text = text.replace(before, after)
    # Rating-only descriptions for advisories without CVE identifiers.
    text = re.sub(r"\bThe advisory rates the issue under CWE-200\b",
                  "The issue is classified under CWE-200", text, flags=re.IGNORECASE)
    text = re.sub(r"\bThe advisory rates this (?:critical|high|medium|low) severity\b",
                  "The advisory describes this", text, flags=re.IGNORECASE)
    text = re.sub(r"\bThe advisory rates this as authenticated remote code execution\b",
                  "The advisory describes authenticated remote code execution", text, flags=re.IGNORECASE)
    # Unfinished clause after rating removal has no technical meaning.
    text = re.sub(r",\s*a\.$", ".", text)
    text = re.sub(r"\s{2,}", " ", text).strip()
    df_clean.at[idx, "model_text_final"] = text

# Disallow CVSS vectors and direct label disclosures in the final model text.
assert not df_clean["model_text_final"].str.contains(
    r"\bAV:[NALP]/AC:[LHF]|\bThe advisory rates this (?:critical|high|medium|low)\b",
    case=False, regex=True
).any()

changed_final = df_clean["model_text_final"].ne(df_clean["model_text_v7"])
assert df_clean["model_text_final"].str.len().gt(0).all()
assert not df_clean["model_text_final"].str.contains(identifier_pattern, case=False, regex=True).any()
assert not df_clean["model_text_final"].str.contains(
    r"\bIt has been declared as (?:critical|high|medium|low)\b|"
    r"\bThe source rates the severity (?:critical|high|medium|low)\b|"
    r"\b(?:base|estimated) score .*?\d\.\d\s*\((?:critical|high|medium|low)\)",
    case=False, regex=True
).any()
print("Residual-leakage cleanup modified records:", changed_final.sum())
print("Remaining empty model texts:", (df_clean["model_text_final"] == "").sum())
print("Direct-rating audit passed; original labels and split retained.")
```
